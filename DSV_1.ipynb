# List


# Defination

### List is a collection of ordered items stored in square bracket [ ]
### it is mutable ( elements can be changed )

```python
Cities = ["Delhi" , "Banglore" , "Bihar"]
```

```python
# create one more list and just print it

marks = [45,35,24,86,54,85]
print(marks)
```

```python
# list with different data types

student = [420,"Raj","CSE",8.79]
print(student)
```

```python
# Access element by index
student = [345,"Amanda","AIDS",9.93]
print(student[2])
print(student[-3])
```

```python
# changing an element in the string
cgpa = [6.4,8.5,3.6,7.4]
print(cgpa)
cgpa[2] = 9.5
print("After Update")
print(cgpa)
```

```python
# add an element in the list by using append() function
student = ["Raaj","Mayank","Rahul","Rana"]
print(student)
student.append("Bhavesh")
print(student)
```

```python
# add an item in a particular location
student = ["Raaj","Mayank","Rahul","Rana"]
print(student)
student.insert(2,"Bhavesh")
print(student)
```

```python
# Extend two list
section_a = ["Rahul","Raj","Cheeta"]
section_b = ["King","Badsha","Farhaan"]
print(section_a)
print(section_b)
section_a.extend(section_b)
print(section_a)
```

```python
# remove an item in a list

subject = ["Python","JAVA","C++","DBMS"]
remove = subject.pop()
print(subject)
print(remove)
```

```python
# to find the length of the list
subject = ["Python","JAVA","C++","DBMS"]
print(len(subject))
```

```python
# Sorting the elements the list
marks = [45,76,23,98,54]
print(marks)
marks.sort()
marks.sort(reverse = True)
print(marks)
```

```python
# nested list
marks = [
    [47,57,97],
    [78,68,76],
    [65,75,25]
]
print(marks)
print(marks[0])
print(marks[0][1])
print(marks[2][2])
```

```python
# create a student record managment
# create a list with student id , student name , dept , list of subjects , list of marks
  # display all the items in the list
  # arrange the subject names in ascending order
  # find the maxium marks and display
  # display miniimum marks
  # display average marks
```

```python
student = [
    420,
    "Raj",
    "CSE",
    ["JAVA" , "Python" , "DBMS" , "DSA"],
    [78 , 54 , 95 ,24],
]

print("Student_id = ",student[0])
print("Student_name = ",student[1])
print("Student_dept = ",student[2])
print("Student_subject = ",student[3])
print("Student_marks = ",student[4])

student[3].sort()
print("Student_subject_sort = ",student[3])

maximum = max(student[4])
print("Student_maximum_marks = ",maximum)

minimum = min(student[4])
print("Student_minimum_marks = ",minimum)

average = sum(student[4]) / len(student[4])
print("Student_average_marks = ",average)
```

# Tuple

# defination

#### A tuple is ordered collection of items

####It is immutable
####Tuple given in the ()

```python
# create a tuple

student = ("Raj",23,'CSE',8.9)
print(student)
```

```python
# display the items using the index

Subjects = ('Java','DBMS','DSA','Python')
print(Subjects[0])
print(Subjects[1])
print(Subjects[2])
print(Subjects[3])
```

```python
Subjects = ('Java','DBMS','DSA','Python')
Subjects[2] = 'C++'
print(Subjects)
```

```python
# tuple slicing and print the length , find the item is present in tuple

student = ("Raj",23,'CSE',8.9)
print(student[0:2])
print("Number of subjects = ",len(student))
print("Deep learning " in student)
print("CSE" in student)
```

```python
# nested tuple
student = (420 , 'Raj' , (78 , 45 , 85) , 'CSE')
print(student)
print(student[2])
print(student[2][1])
```
