# Tuple

### Ordered collection of items ()
### One main difference is that tuple is immutable

##var_name = list[[stu1:()],[stu2:()]...]

```python
# Create a Tuple
student = (101,"Raj","CSE",8.9)
print(student)
# tuples using index
print("Student Information in first location ",student[0])
print(student[1])
# negative index
print("Example for negative index ",student[-2])
```

```python
# To find the length of the tuple
subject = ('DSA','DSV','OS','Python','ML')
print(len(subject))
```

```python
# Slicing
subjects = ('DSA','DSV','OS','Python','ML')
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
# Tuple Immutable
subjects = ('DSA','DSV','OS','Python','ML')
subjects[2] = 'Java'
```

```python
# Nested Tuple
student = (101,'Raj',(87,43,87))
print(student)
print(student[1])
print(student[2])
print(student[2][1])
```

```python
# Count Function
marks = (45,54,54,77,98)
print("Time of Occurance " , marks.count(54))
```

```python
# Find the element present in tuple or not
subjects = ('DSA','DSV','OS','Python','ML')
print('DSA' in subjects)
print('Java' in subjects)
```

# SET

## Unordered collection of items
## mutable
## allows the diferent data types
## it dosen't support for index
## it gives output with uniques (NO Duplicate)

```python
# Symbol for set { }
# Create a Set with Similar items
marks = {75,68,89,55,78,89,75,66,55,99}
print(marks)
```

```python
# Create a empty set
a = {}
a = set()
# check its type
print(type(a))
```

```python
# Set dosen't support the index
students={'Amit','Raj','Mayank'}
print(students[1])
```

```python
# Add an item in the set
students={'Amit','Raj','Mayank'}
students.add('Raju')
print(students)

# To Update multiple item in a set
students.update(['Ram','Sita','Dev','Vani'])
print(students)
```

```python
# Remove an item from the set
students = {'Amit', 'Raj', 'Mayank'}
students.remove('Raj')
print(students)


students.discard('Mayank')
print(students)
```

```python
students = {'Amit','Mayank','Raj'}
students.remoove('Raju')
print(students)
```

```python
students = {'Amit','Mayank','Raj'}
students.discard('Raju')
print(students)
```

```python
# Create a set and clear list of items
students_info = {
    'Amit',
    'Raj',
    'Mayank'
}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks = {99,74,84,95}
del marks
print(marks)
```

```python
# set operations
python_students = {'Amin','Neha','Sanjiv','Manish'}
java_students = {'Amit','Karan','Dhillip','Manish','Neha'}

# For union operations by using set
result = python_students.union(java_students)
print("Union Operations",result)

# for intersection operation by using set
result = python_students.intersection(java_students)
print("Intersection Operation",result)

# for difference operation by using set
result = python_students.difference(java_students)
print("For Difference",result)
```

```python
# Converting list to set
stu_id = [101,102,103,104,105]
unique_stud_id = set(stu_id)
print(unique_stud_id)
unique_roll_no = list(unique_stud_id)
print(unique_roll_no)
```
